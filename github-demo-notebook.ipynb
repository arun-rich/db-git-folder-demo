```python
items=dbutils.fs.ls('/databricks-datasets/')
directories=[]
files=[]
for directory in items:
  if directory.isDir():
    directories.append(directory.path)
  else:
      files.append(directory.path)
print('directories')
print('--------------')
print(directories)
print('Files')
print('--------------')
print(files)
```
